```python
import pandas as pd

# Sample dataset
data = {
    'item_id': [1, 2, 3, 4, 5],
    'title': ['The Matrix', 'Inception', 'Interstellar', 'The Godfather', 'Pulp Fiction'],
    'genres': ['Action, Sci-Fi', 'Action, Sci-Fi', 'Adventure, Drama', 'Crime, Drama', 'Crime, Drama'],
    'description': [
        'A computer hacker learns about the true nature of his reality.',
        'A thief who steals corporate secrets is given the inverse task.',
        'A team of explorers travels through a wormhole in space.',
        'The aging patriarch of an organized crime dynasty transfers control.',
        'The lives of two mob hitmen, a boxer, and a gangster\'s wife intertwine.'
    ]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Combine genres and descriptions into a single string
df['combined'] = df['genres'] + ' ' + df['description']

# Vectorization
count_vectorizer = CountVectorizer()
count_matrix = count_vectorizer.fit_transform(df['combined'])

# Compute cosine similarity matrix
cosine_sim = cosine_similarity(count_matrix, count_matrix)
```

```python
def recommend(title, cosine_sim=cosine_sim):
    # Get the index of the item that matches the title
    idx = df[df['title'] == title].index[0]
    
    # Get the pairwise similarity scores of all items with that item
    sim_scores = list(enumerate(cosine_sim[idx]))
    
    # Sort the items based on the similarity scores
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    
    # Get the scores of the top 3 most similar items
    sim_scores = sim_scores[1:4]
    
    # Get the item indices
    item_indices = [i[0] for i in sim_scores]
    
    # Return the top 3 most similar items
    return df['title'].iloc[item_indices]

# Example usage
recommended_items = recommend('The Matrix')
print("Recommended items for 'The Matrix':")
print(recommended_items)
```
