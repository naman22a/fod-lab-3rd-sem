# Data Visualization

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(15, 5))

# Python scores
plt.subplot(1, 3, 1)  # 1 row, 3 columns, 1st subplot
plt.hist(df['python'], bins=5, color='yellow', edgecolor='black')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.title('Histogram of Python Scores')

# Java scores
plt.subplot(1, 3, 2)  # 1 row, 3 columns, 2nd subplot
plt.hist(df['java'], bins=5, color='red', edgecolor='black')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.title('Histogram of Java Scores')

# C scores
plt.subplot(1, 3, 3)  # 1 row, 3 columns, 3rd subplot
plt.hist(df['c'], bins=5, color='blue', edgecolor='black')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.title('Histogram of C Scores')

plt.show()
```
