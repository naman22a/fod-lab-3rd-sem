# Create synthetic dataset

```python
import pandas as pd
from faker import Faker
```

```python
fake = Faker()
```

```python
def generate_data(num_rows):
    data = []
    for _ in range(num_rows):
        data.append({
            'ID': fake.unique.random_number(digits=5),
            'Name': fake.name(),
            'Age': fake.random_int(min=18, max=80),
            'Country': fake.country(),
            'Salary': fake.random_int(min=30000, max=120000)
        })
    return data
```

```python
num_rows = 100
df = pd.DataFrame(generate_data(num_rows))
df.to_csv('09-synthetic_dataset.csv', index=False)
```
